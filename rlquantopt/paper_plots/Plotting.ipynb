# Potential wells

```python
# Define x values and cosine curve
x = np.linspace(-np.pi, np.pi, 400)
y = -np.cos(x)  # Negative cosine to create a well shape

comp_bg_c = 'lightblue'
comp_fg_c = 'b'
leak_bg_c = 'pink'
leak_fg_c = 'r'
# Define quantum states and their y positions
states = {
    '|0⟩': -0.7,
    '|1⟩': -0.4,
    '|2⟩': 0.0,
    '|3⟩': 0.3,
    # '|4⟩': 0.6
}
comp_leak_boundary = -0.2

# Create figure and axes
fig, ax = plt.subplots(figsize=(4, 4))

# Plot cosine curve
ax.plot(x, y, 'k', linewidth=2)  # Black curve

# Shade computational and leakage regions
ax.fill_between(x, -1.2, comp_leak_boundary, color=comp_bg_c)#, alpha=0.5)
ax.fill_between(x, comp_leak_boundary, 1.1, color=leak_bg_c)#, alpha=0.5)

# Add horizontal lines for quantum states spanning to the cosine curve
for state, y_pos in states.items():
    # Find intersection points with the cosine curve
    left_x = x[np.argmin(np.abs(y - y_pos))]  # Closest x where cosine is at y_pos
    right_x = -left_x  # Symmetric for cosine curve
    
    # Determine color (blue for computational, red for leakage)
    color_bg = leak_bg_c if y_pos > comp_leak_boundary else comp_bg_c
    color_fg = leak_fg_c if y_pos > comp_leak_boundary else comp_fg_c
    
    # Draw the energy level line
    ax.hlines(y=y_pos, xmin=left_x, xmax=right_x, color=color_fg, linewidth=2)
    
    # Label the quantum state slightly to the right
    ax.text((left_x + right_x)/2, y_pos, state, fontsize=12, horizontalalignment='center',verticalalignment='center', color=color_fg, bbox=dict(ec='None', fc=color_bg))

# Labels for regions
ax.text(0, -1.15, 'Computational', fontsize=14, color=comp_fg_c, ha='center')
ax.text(0, 0.7, 'Leakage', fontsize=14, color=leak_fg_c, ha='center')

# Add dotted line ("...")
ax.text(0, 0.45, '⋮', fontsize=20, color='red', ha='center')  # Unicode vertical dots

# Remove ticks and labels
ax.set_xticks([])
ax.set_yticks([])
ax.set_frame_on(False)

# Show the figure
plt.show()
```
